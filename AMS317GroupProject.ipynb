```python
#Hypothesis 1: Using logistic regression, ROC/AUC evaluation, and interaction modeling to assess lifestyle factors affecting obesity.

!pip install ucimlrepo -q

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor

from sklearn.metrics import roc_curve, auc, confusion_matrix, classification_report

from ucimlrepo import fetch_ucirepo

dataset = fetch_ucirepo(id=544)

X = dataset.data.features
y = dataset.data.targets

df = pd.concat([X, y], axis=1)

print("Raw Data Preview (from ucimlrepo):")
display(df.head())
print("Shape:", df.shape)

favc = df["FAVC"].astype(str).str.strip().str.lower()
df["FAVC_bin"] = np.where(favc == "yes", 1,
                   np.where(favc == "no", 0, np.nan))

df["NCP_num"] = pd.to_numeric(df["NCP"], errors="coerce")

labels = df["NObeyesdad"].astype(str).str.strip().str.lower()

under_normal = ["insufficient_weight", "normal_weight"]
over_obese = [
    "overweight_level_i", "overweight_level_ii",
    "obesity_type_i", "obesity_type_ii", "obesity_type_iii"
]

df["OverObese"] = np.where(labels.isin(over_obese), 1,
                    np.where(labels.isin(under_normal), 0, np.nan))

df_clean = df[["FAVC_bin", "NCP_num", "OverObese", "NObeyesdad"]].dropna().copy()

print("\nDATA CLEANING SUMMARY:")
print("Rows original:", len(df))
print("Rows used:", len(df_clean))
print("Rows dropped:", len(df) - len(df_clean))

print("\nDescriptive Statistics for NCP_num (meals per day):")
display(df_clean["NCP_num"].describe())

print("\nDistribution of Weight Categories (NObeyesdad):")
display(df_clean["NObeyesdad"].value_counts())

print("\nBinary Outcome Distribution (OverObese):")
display(df_clean["OverObese"].value_counts())

ct = pd.crosstab(df_clean["FAVC_bin"], df_clean["OverObese"], margins=True)
ct.index = ["FAVC=No", "FAVC=Yes", "Total"]
ct.columns = ["Not Over/Obese", "Over/Obese", "Row Total"]

print("\nContingency Table: FAVC vs OverObese:")
display(ct)

plt.figure(figsize=(6,4))
plt.hist(df_clean["NCP_num"], bins=10, edgecolor="black")
plt.title("Histogram of NCP (Number of Main Meals)")
plt.xlabel("NCP (meals per day)")
plt.ylabel("Frequency")
plt.show()

prop_over = pd.crosstab(df_clean["FAVC_bin"], df_clean["OverObese"], normalize="index")

plt.figure(figsize=(6,4))
plt.bar(["FAVC=No", "FAVC=Yes"],
        [prop_over.loc[0,1], prop_over.loc[1,1]])
plt.title("Proportion Overweight/Obese by FAVC")
plt.ylabel("Proportion Overweight/Obese")
plt.ylim(0,1)
plt.show()

import numpy as np
import matplotlib.pyplot as plt
x = df_clean["NCP_num"]
y = df_clean["OverObese"]
formula = "OverObese ~ FAVC_bin + NCP_num + FAVC_bin:NCP_num"
model = smf.logit(formula=formula, data=df_clean).fit(disp=False)
np.random.seed(42)
y_jitter = y + np.random.uniform(-0.02, 0.02, size=len(y))



print("\nLOGISTIC REGRESSION RESULTS:")
display(model.summary2())

params = model.params
conf = model.conf_int()

or_table = pd.DataFrame({
    "Odds_Ratio": np.exp(params),
    "CI_lower": np.exp(conf[0]),
    "CI_upper": np.exp(conf[1]),
    "p_value": model.pvalues
})

print("\nODDS RATIOS WITH 95% CI:")
display(or_table)

X_vif = pd.DataFrame({
    "Intercept": 1.0,
    "FAVC_bin": df_clean["FAVC_bin"].astype(float),
    "NCP_num": df_clean["NCP_num"].astype(float),
    "Interaction": (df_clean["FAVC_bin"] * df_clean["NCP_num"]).astype(float)
})

vif_df = pd.DataFrame({
    "Variable": X_vif.columns,
    "VIF": [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
})

print("\nVIF (Multicollinearity Check):")
display(vif_df)

fit_stats = {
    "Log-Likelihood": model.llf,
    "AIC": model.aic,
    "BIC": model.bic,
    "McFadden_R2": model.prsquared
}
fit_df = pd.DataFrame.from_dict(fit_stats, orient="index", columns=["Value"])

print("\nMODEL FIT STATISTICS:")
display(fit_df)

interpretations = []
for name, row in or_table.iterrows():
    if name == "Intercept":
        continue
    OR = row["Odds_Ratio"]
    p = row["p_value"]
    ci_l = row["CI_lower"]
    ci_u = row["CI_upper"]
    sig = "statistically significant" if p < 0.05 else "NOT statistically significant"
    direction = "higher" if OR > 1 else "lower"
    interpretations.append(
        f"{name}: OR = {OR:.2f} (95% CI [{ci_l:.2f}, {ci_u:.2f}]), "
        f"p = {p:.4f}. This predictor is {sig} and indicates {direction} odds of being overweight/obese."
    )

print("\nPLAIN-ENGLISH INTERPRETATION:")
for line in interpretations:
    print("-", line)

y_true = df_clean["OverObese"]
y_score = model.predict(df_clean)

# -------------------------------
# S-shaped logistic curve plot
# -------------------------------

# Linear predictor (logit) for each observation: Xβ
X_design = model.model.exog            # design matrix used by statsmodels
lin_pred = X_design @ model.params     # linear predictor

# Sort by linear predictor so the curve is monotone
order = np.argsort(lin_pred)
lin_sorted = lin_pred[order]
prob_sorted = 1 / (1 + np.exp(-lin_sorted))   # logistic transformation
y_sorted = y_true.iloc[order].values

plt.figure(figsize=(6, 4))

# Logistic S-curve
plt.plot(lin_sorted, prob_sorted, label="Fitted logistic curve")

# "Rug" marks for actual outcomes (0 vs 1)
plt.scatter(lin_sorted[y_sorted == 0],
            np.zeros((y_sorted == 0).sum()) + 0.02,
            marker="|", s=30, label="Observed 0")
plt.scatter(lin_sorted[y_sorted == 1],
            np.ones((y_sorted == 1).sum()) - 0.02,
            marker="|", s=30, label="Observed 1")

plt.ylim(-0.05, 1.05)
plt.xlabel("Linear predictor (log-odds)")
plt.ylabel("Predicted probability of OverObese = 1")
plt.title("Logistic Regression Fitted Curve with Observed Outcomes")
plt.legend(loc="lower right")
plt.show()

fpr, tpr, thresholds = roc_curve(y_true, y_score)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6, 4))
plt.plot(fpr, tpr, label=f"ROC curve (AUC = {roc_auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--", label="Random guess")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve for OverObese Classification")
plt.legend(loc="lower right")
plt.show()

print(f"\nAUC: {roc_auc:.3f}")

threshold = 0.5
y_pred = (y_score >= threshold).astype(int)

cm = confusion_matrix(y_true, y_pred)
cm_df = pd.DataFrame(
    cm,
    index=["Actual: Not Over/Obese", "Actual: Over/Obese"],
    columns=["Predicted: Not Over/Obese", "Predicted: Over/Obese"]
)

print("\nConfusion Matrix (threshold = 0.5):")
display(cm_df)

print("\nClassification Report (threshold = 0.5):")
print(classification_report(y_true, y_pred, target_names=["Not Over/Obese", "Over/Obese"]))

ncp_grid = np.linspace(df_clean["NCP_num"].min(), df_clean["NCP_num"].max(), 50)

grid_no_favc = pd.DataFrame({"FAVC_bin": 0, "NCP_num": ncp_grid})
grid_yes_favc = pd.DataFrame({"FAVC_bin": 1, "NCP_num": ncp_grid})

pred_no_favc = model.predict(grid_no_favc)
pred_yes_favc = model.predict(grid_yes_favc)

plt.figure(figsize=(6, 4))
plt.plot(ncp_grid, pred_no_favc, label="FAVC = No (does NOT often eat high-calorie foods)")
plt.plot(ncp_grid, pred_yes_favc, label="FAVC = Yes (often eats high-calorie foods)")
plt.xlabel("NCP (Number of Main Meals per Day)")
plt.ylabel("Predicted Probability of Overweight/Obese")
plt.title("Predicted Probability of Overweight/Obese by NCP and FAVC")
plt.ylim(0, 1)
plt.legend()
plt.show()
```

```python
!pip install ucimlrepo -q
import pandas as pd
import numpy as np
from ucimlrepo import fetch_ucirepo


dataset = fetch_ucirepo(id=544)
X = dataset.data.features
y = dataset.data.targets
df = pd.concat([X, y], axis=1)

# Computing BMI
df["BMI"] = df["Weight"] / (df["Height"] ** 2)

# Converting categorical variables
df["NObeyesdad"] = df["NObeyesdad"].astype("category")
df["MTRANS"] = df["MTRANS"].astype("category")
```

```python
print("\n=== Descriptive Stats ===")
print(df[["FAF", "BMI"]].describe())

print("\n=== Transportation Mode Counts ===")
print(df["MTRANS"].value_counts())

print("\n=== Obesity Level Counts ===")
print(df["NObeyesdad"].value_counts())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x="FAF", y="BMI", alpha=0.6)
plt.title("Relationship Between Physical Activity Frequency (FAF) and BMI")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x="MTRANS", y="BMI")
plt.xticks(rotation=45)
plt.title("BMI Differences Across Transportation Modes")
plt.show()
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

model_anova = ols("BMI ~ C(MTRANS)", data=df).fit()
anova_table = sm.stats.anova_lm(model_anova, typ=2)
print("\n=== ANOVA: BMI ~ Transportation Mode ===")
print(anova_table)
```

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
import seaborn as sns
import matplotlib.pyplot as plt

model_faf = ols("BMI ~ FAF", data=df).fit()
print("\n=== Regression: BMI ~ FAF ===")
print(model_faf.summary())

plt.figure(figsize=(8,6))


sns.scatterplot(
    data=df,
    x="FAF",
    y="BMI",
    alpha=0.6
)


sns.regplot(
    data=df,
    x="FAF",
    y="BMI",
    scatter=False,
    ci=None,
    line_kws={'color': 'red', 'linewidth': 2}
)

plt.title("Regression of BMI on Physical Activity Frequency (FAF)")
plt.xlabel("Physical Activity Frequency (FAF)")
plt.ylabel("BMI")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df["BMI"], kde=True, bins=30)
plt.title("Distribution of BMI Values")
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
